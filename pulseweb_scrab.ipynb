```python

print("Page title:")
print(soup.title.get_text(strip=True) if soup.title else "Not found")

print("\nMetadata:")
for tag in soup.find_all("meta"):
    name = tag.get("name") or tag.get("property")
    content = tag.get("content")

    if name and content:
        print(name, ":", content)
```
